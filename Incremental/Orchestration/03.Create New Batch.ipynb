```python
from pyspark.sql import Row
from pyspark.sql.functions import current_timestamp
```

```python
dbutils.widgets.text("p_batch_id", "")
v_batch_id = dbutils.widgets.get("p_batch_id")
```

```python
control_table = "formula1_inc.control.batch_control"
```

```python
if v_batch_id:
    input_df = (
        spark.createDataFrame([Row(batch_id = v_batch_id,status = "in_progress")])
             .withColumn("created_timestamp",current_timestamp())
             .withColumn("updated_timestamp",current_timestamp())
    )
    input_df.write.mode("append").saveAsTable(control_table)
    print(f"Batch id {v_batch_id} is in progress")
else:
    raise Exception("there is no batch to process")
```
