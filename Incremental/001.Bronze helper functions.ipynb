```python
def add_ingestion(df):
    return (
        df.withColumn("ingestion_time",current_timestamp())
          .withColumn("filename",col("_metadata.file_path"))
          .withColumn("batch_id",lit(v_batch_id))
    )
```

```python
def write_to_bronze(input_df,table_name):
    return(
        input_df.write.mode("overwrite")
                      .partitionBy("batch_id")
                      .option("replaceWhere",f"batch_id = '{v_batch_id}'")
                      .format("delta")
                      .saveAsTable(table_name)
    )
```
