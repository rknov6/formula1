```sql
create or replace view formula1.gold.v_constructors_standings
as
with cte as (
select
s.season,
c.constructor_id,
c.constructor_name,
c.nationality,
sum(s.points) as total_points,
count_if(s.is_win) as number_of_wins,
count_if(s.is_podium) as number_of_podiums
from formula1.gold.fact_session_results s
join formula1.gold.dim_constructors c on s.constructor_id = c.constructor_id
group by all
)
select 
dense_rank() over (partition by season order by total_points desc,number_of_wins desc,number_of_podiums desc) as position,
*
from cte
```
