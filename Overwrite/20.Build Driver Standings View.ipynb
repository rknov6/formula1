```sql
create or replace view formula1.gold.v_driver_standings
as
with cte as (select s.season,
       d.driver_id,
       d.driver_name,
       d.nationality,
       count(*) as race_starts,
       sum(s.points) as total_points,
       count_if(s.is_win) as number_of_wins,
       count_if(s.is_podium) as number_of_podiums
from formula1.gold.dim_drivers d 
    join formula1.gold.fact_session_results s
    on s.driver_id = d.driver_id
group by all
)
select 
dense_rank() over (partition by season order by total_points desc,number_of_wins desc) as position,
*
from cte
```
